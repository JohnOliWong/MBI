```python
import torch
import numpy as np
from Dataloader import Dataset, Load_Dataset_A, Load_Dataset_B, Load_Dataset_C
from model import fNIRS_T, fNIRS_PreT
from Parameters import parameters
from LOSO_Split import Split_Dataset_A, Split_Dataset_B, Split_Dataset_C
import os
from sklearn.metrics import precision_score, recall_score, f1_score, cohen_kappa_score
import pandas as pd

class LabelSmoothing(torch.nn.Module):
    """NLL loss with label smoothing."""
    def __init__(self, smoothing=0.0):
        super(LabelSmoothing, self).__init__()
        self.confidence = 1.0 - smoothing
        self.smoothing = smoothing

    def forward(self, x, target):
        logprobs = torch.nn.functional.log_softmax(x, dim=-1)
        nll_loss = -logprobs.gather(dim=-1, index=target.unsqueeze(1))
        nll_loss = nll_loss.squeeze(1)
        smooth_loss = -logprobs.mean(dim=-1)
        loss = self.confidence * nll_loss + self.smoothing * smooth_loss
        return loss.mean()


if __name__ == "__main__":
    # Training epochs
    EPOCH = 200

    data_path, dataset, model = parameters()

    # Save file and avoid training file overwriting.
    save_path = 'Results/' + dataset + '/LOSO/' + model
    assert os.path.exists(save_path) is False, 'path is exist'
    os.makedirs(save_path)
    log_excel = os.path.join(save_path, 'Log.xlsx')

    # Load dataset, set flooding levels and number of Subjects. Different models may have different flooding levels.
    if dataset == 'A':
        flooding_level = [0, 0, 0]
        Subjects = 8
        if model == 'fNIRS-T':
            feature, label = Load_Dataset_A(data_path, model='fNIRS-T')
        elif model == 'fNIRS-PreT':
            feature, label = Load_Dataset_A(data_path, model='fNIRS-PreT')
    elif dataset == 'B':
        if model == 'fNIRS-T':
            flooding_level = [0.45, 0.40, 0.35]
        else:
            flooding_level = [0.40, 0.38, 0.35]
        Subjects = 29
        feature, label = Load_Dataset_B(data_path)
    elif dataset == 'C':
        flooding_level = [0.45, 0.40, 0.35]
        Subjects = 30
        feature, label = Load_Dataset_C(data_path)

    _, _, channels, sampling_points = feature.shape

    for sub in range(1, Subjects+1):
        # Spilt dataset, one subject's data is test set, and the rest is training set.
        if dataset == 'A':
            X_train, y_train, X_test, y_test = Split_Dataset_A(sub, feature, label, channels)
        elif dataset == 'B':
            X_train, y_train, X_test, y_test = Split_Dataset_B(sub, feature, label, channels)
        elif dataset == 'C':
            X_train, y_train, X_test, y_test = Split_Dataset_C(sub, feature, label, channels)

        path = save_path + '/' + str(sub)
        assert os.path.exists(path) is False, 'sub-path is exist'
        os.makedirs(path)

        train_set = Dataset(X_train, y_train, transform=True)
        test_set = Dataset(X_test, y_test, transform=True)
        train_loader = torch.utils.data.DataLoader(train_set, batch_size=128, shuffle=True)
        test_loader = torch.utils.data.DataLoader(test_set, batch_size=128, shuffle=True)

        # -------------------------------------------------------------------------------------------------------------------- #
        device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
        if dataset == 'A':
            if model == 'fNIRS-T':
                net = fNIRS_T(n_class=2, sampling_point=sampling_points, dim=64, depth=6, heads=8, mlp_dim=64).to(device)
            elif model == 'fNIRS-PreT':
                net = fNIRS_PreT(n_class=2, sampling_point=sampling_points, dim=64, depth=6, heads=8, mlp_dim=64).to(device)
        elif dataset == 'B':
            if model == 'fNIRS-T':
                net = fNIRS_T(n_class=2, sampling_point=sampling_points, dim=64, depth=6, heads=8, mlp_dim=64).to(device)
            elif model == 'fNIRS-PreT':
                net = fNIRS_PreT(n_class=2, sampling_point=sampling_points, dim=64, depth=6, heads=8, mlp_dim=64).to(device)
        elif dataset == 'C':
            if model == 'fNIRS-T':
                net = fNIRS_T(n_class=3, sampling_point=sampling_points, dim=128, depth=6, heads=8, mlp_dim=64).to(device)
            elif model == 'fNIRS-PreT':
                net = fNIRS_PreT(n_class=3, sampling_point=sampling_points, dim=128, depth=6, heads=8, mlp_dim=64).to(device)

        criterion = LabelSmoothing(0.1)
        optimizer = torch.optim.AdamW(net.parameters())
        lrStep = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=30)
        # -------------------------------------------------------------------------------------------------------------------- #
        test_max_acc = 0
        eval_loss = []
        all_labels = []
        all_preds = []
        acc_list = []
        for epoch in range(EPOCH):
            net.train()
            train_running_acc = 0
            total = 0
            loss_steps = []
            
            for i, data in enumerate(train_loader):
                inputs, labels = data
                inputs = inputs.to(device)
                labels = labels.to(device)
                outputs = net(inputs)
                loss = criterion(outputs, labels.long())

                if epoch < 30:
                    b = flooding_level[0]
                elif epoch < 50:
                    b = flooding_level[1]
                else:
                    b = flooding_level[2]
                loss = (loss - b).abs() + b

                optimizer.zero_grad()
                loss.backward()
                optimizer.step()

                loss_steps.append(loss.item())
                total += labels.shape[0]
                preds = outputs.argmax(dim=1, keepdim=True)
                train_running_acc += preds.eq(labels.view_as(preds)).sum().item()

            train_running_loss = float(np.mean(loss_steps))
            train_running_acc = 100 * train_running_acc / total
            print(f"[{sub}, {epoch}] Train loss: {train_running_loss:.2f}, Train acc: {train_running_acc:.2f}%")

            net.eval()
            correct = 0
            total = 0
            eval_loss = []

            with torch.no_grad():
                for data in test_loader:
                    inputs, labels = data
                    inputs = inputs.to(device)
                    labels = labels.to(device)
                    outputs = net(inputs)
                    loss = criterion(outputs, labels.long())

                    eval_loss.append(loss.item())
                    total += labels.shape[0]
                    preds = outputs.argmax(dim=1, keepdim=True)
                    correct += preds.eq(labels.view_as(preds)).sum().item()
                    all_labels.append(labels.cpu().numpy())
                    all_preds.append(preds.cpu().numpy())

            # compute evaluation accuracy
            acc_mean = 100 * correct / total
            acc_list.append(acc_mean)
            test_running_loss = float(np.mean(eval_loss))
            print(f"[{sub}] Test loss: {test_running_loss:.2f}, Test acc: {acc_mean:.2f}%")

            # update learning rate
            lrStep.step()

        # performance evaluation
        all_labels = np.concatenate(all_labels).reshape(-1)
        all_preds = np.concatenate(all_preds).reshape(-1)

        acc_mean, acc_std = np.mean(acc_list[-50:]), np.std(acc_list[-50:])
        precision_per_class = precision_score(all_labels, all_preds, average=None)
        recall_per_class = recall_score(all_labels, all_preds, average=None)
        f1_per_class = f1_score(all_labels, all_preds, average=None)
        kappa = cohen_kappa_score(all_labels[-50:], all_preds[-50:])

        precision_mean, precision_std = 100 * np.mean(precision_per_class[-50:]), 100 * np.std(precision_per_class[-50:])
        recall_mean, recall_std = 100 * np.mean(recall_per_class[-50:]), 100 * np.std(recall_per_class[-50:])
        f1_mean, f1_std = 100 * np.mean(f1_per_class[-50:]), 100 * np.std(f1_per_class[-50:])

        # save the results
        with open(path + '/Metrics.txt', 'w') as f:
            f.write(f'Accuracy: {acc_mean:.2f} ± {acc_std:.2f}\n')
            f.write(f'Precision: {precision_mean:.2f} ± {precision_std:.2f}\n')
            f.write(f'Recall: {recall_mean:.2f} ± {recall_std:.2f}\n')
            f.write(f'F1-Score: {f1_mean:.2f} ± {f1_std:.2f}\n')
            f.write(f'Kappa: {kappa:.2f}\n')
        
        df = pd.DataFrame([[acc_mean, acc_std, precision_mean, precision_std, recall_mean, recall_std, f1_mean, f1_std, kappa]],
                        columns=['Accuracy', 'Std', 'Precision', 'Std', 'Recall', 'Std', 'F1', 'Std', 'Kappa'])

        if not os.path.exists(log_excel):
            df.to_excel(log_excel, index=False, engine='openpyxl')
        else:
            with pd.ExcelWriter(log_excel, mode='a', engine='openpyxl', if_sheet_exists='overlay') as writer:
                if 'Sheet1' in writer.sheets:
                    startrow = writer.sheets['Sheet1'].max_row
                else:
                    startrow = 0
                df.to_excel(writer, index=False, header=False, startrow=startrow)
```
